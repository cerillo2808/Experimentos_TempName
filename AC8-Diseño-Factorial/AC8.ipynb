```r
yield = read.csv("FactorialDesign.csv",
                 header = T,
                 encoding = "UTF-8")

# Debemos quitar el núnero de corrida para no crear interacciones espurias
yield <- yield[-1] 
plot(yield)

par(mfrow=c(2,2))
interaction.plot(yield$Temp,yield$Conc,yield$Yield)
interaction.plot(yield$Temp,yield$Catalyst,yield$Yield)
interaction.plot(yield$Conc,yield$Catalyst,yield$Yield)
par(mfrow=c(1,1))

# Modelamos los efectos principales
lm1 = lm(Yield ~ ., data = yield)
summary(lm1)

# Aquí se realizan o agregan las interacciones de 2 factores
lm2 = lm(Yield ~ .^2, data = yield)
summary(lm2)

#compare the two models
anova(lm1, lm2)

# En este paso realizamos las interacciones del modelo completo
lm3 = lm(Yield ~ .^3, data = yield)
summary(lm3)

# Ojo, los resultados del summary(lm3) son algo distintos verdad. Esto es porque en este
# experimento no tenemos réplicas, sino que se realizaron corridas con distinta Temp. Y no
# podemos analizar la significancia de de los coeficientes del modelo complete. Recuerden el
# problema del 3k – se pueden volver intratable
# Con base en modelos y resultados anteriores, sin embargo, podemos modelar usando las
# interacciones más significativas
lm4 = lm(Yield ~ Conc + Temp*Catalyst, data = yield)
summary(lm4)  
```

## Ejercicio 2. Comparación de los modelos


```r
comparacion = data.frame(
  Modelo = c("Efectos principales", "Interacciones de dos factores", "Modelo completo"),
  R2_ajustado = c(summary(lm1)$adj.r.squared, summary(lm2)$adj.r.squared, summary(lm3)$adj.r.squared),
  GL_residuales = c(df.residual(lm1), df.residual(lm2), df.residual(lm3))
)
comparacion
```

El modelo lm1 analiza los efectos individuales de Temp, Conc y Catalyst sobre Yield. El modelo lm2 añade las interacciones Temp:Conc, Temp:Catalyst y Conc:Catalyst, lo que permite que el efecto de un factor dependa del nivel de otro.El modelo completo lm3 incorpora también la interacción triple Temp:Conc:Catalyst. El R² ajustado permite comparar la variación de Yield explicada por los modelos, teniendo en cuenta la cantidad de términos. Si el valor del R² ajustado del modelo lm2 es mayor que el del modelo lm1, las interacciones de dos factores aportan capacidad explicativa adicional. En el contexto de la síntesis química, esto indica que el rendimiento no depende solo de cada condición por separado, sino también de la combinación de temperatura, concentración y catalizador. El modelo completo no tiene réplicas por combinación y queda sin grados de libertad residuales para estimar el error. Por eso, su ajuste perfecto a las observaciones no significa que sea un modelo más confiable. Su R² ajustado no puede utilizarse para una comparación convencional.
